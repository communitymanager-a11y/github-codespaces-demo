## Claude Agent SDK を呼び出す

環境変数 `ANTHROPIC_API_KEY` が必要です（Codespaces では作成時に入力した Secret が設定されます）。

```python
from claude_agent_sdk import ClaudeAgentOptions, ResultMessage, query

prompt = "Jupyter Notebook を一文で説明してください。"

async for message in query(prompt=prompt, options=ClaudeAgentOptions(max_turns=1)):
    if isinstance(message, ResultMessage) and message.subtype == "success":
        print(message.result)
```
